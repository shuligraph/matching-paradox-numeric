### Simple Graph

```python
C5 = nx.Graph()
C5.add_edges_from([(1, 2), (2, 3), (3, 4), (4,5), (5,1)])
mu = {1: 0.111, 2: 0.10575, 3: 0.179426, 4: 0.35046, 5: 0.253364} 
F2 = C5.copy(); F2.add_edge(2,5) 
run(C5, mu, label="C5")
run(F2, mu, label="F2")
```

### with loop

```python
G = nx.Graph()
G.add_edges_from([
        (1, 2), (1, 3), (1, 4), (3, 4), (2,2), (3,3), (4,4)
    ])

mu = {1: 0.432685, 2: 0.005861, 3: 0.141128, 4:  0.42032} 

G2 = G.copy(); G2.add_edge(2,3) 

run(G, mu, label="G")
run(G2, mu, label="G2")
```
