```python
color_square = 'tab:blue'
color_triangle = 'tab:orange'

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].scatter(x_square, expected_value_square, marker='o', color=color_square, label='square')
axes[0].scatter(x_square, expected_value_tri, marker='s', color=color_triangle, label='triangle')
axes[0].set_title("Expected Value vs. Number of Nodes")
axes[0].set_xlabel("Number of Nodes")
axes[0].set_ylabel("Expected Value")
axes[0].legend()
axes[0].grid(True)
for x, y in zip(x_square, expected_value_square):
    axes[0].text(x, y, f"{y:.2f}", fontsize=10, va='bottom', ha='right', color=color_square)
for x, y in zip(x_square, expected_value_tri):
    axes[0].text(x, y, f"{y:.2f}", fontsize=10, va='bottom', ha='left', color=color_triangle)

axes[1].semilogy(x_square, total_independent_sets_square, marker='o', color=color_square, label='square')
axes[1].semilogy(x_square, total_independent_sets_tri, marker='s', color=color_triangle, label='triangle')
axes[1].set_title("Total Independent Sets vs. Number of Nodes")
axes[1].set_xlabel("Number of Nodes")
axes[1].set_ylabel("Count of Independent Sets")
axes[1].legend()
axes[1].grid(True)
for x, y in zip(x_square, total_independent_sets_square):
    axes[1].text(x, y, str(int(y)), fontsize=10, va='bottom', ha='right', color=color_square)
for x, y in zip(x_square, total_independent_sets_tri):
    axes[1].text(x, y, str(int(y)), fontsize=10, va='bottom', ha='left', color=color_triangle)

axes[2].semilogy(x_square, total_permutation_sets_square, marker='o', color=color_square, label='square')
axes[2].semilogy(x_square, total_permutation_sets_tri, marker='s', color=color_triangle, label='triangle')
axes[2].set_title("Total Permutation Sets vs. Number of Nodes")
axes[2].set_xlabel("Number of Nodes")
axes[2].set_ylabel("Count of Permutation Sets")
axes[2].legend()
axes[2].grid(True)
for x, y in zip(x_square, total_permutation_sets_square):
    axes[2].text(x, y, str(int(y)), fontsize=10, va='bottom', ha='right', color=color_square)
for x, y in zip(x_square, total_permutation_sets_tri):
    axes[2].text(x, y, str(int(y)), fontsize=10, va='bottom', ha='left', color=color_triangle)

plt.tight_layout()
plt.show()
```
